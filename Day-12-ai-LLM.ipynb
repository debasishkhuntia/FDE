```python

from openai import OpenAI
client = OpenAI(
    api_key="key",
    base_url="https://api.groq.com/openai/v1",
)
def ask_ai(ask):
    response = client.responses.create(
    input=ask,
    model="openai/gpt-oss-20b",
)
    return response.output_text
result=ask_ai("why you?")
print(result)
```
